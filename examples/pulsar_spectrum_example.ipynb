# Pulsar Spectrum Example

This notebook demonstrates how to access pulsar flux measurements and datasets using the FEUPY pulsar spectra interface.

The pulsar spectra data are based on the repository:

**pulsar_spectra — A pulsar flux density catalogue and spectrum fitting repository**

https://all-pulsar-spectra.readthedocs.io/en/latest/

In the following examples we retrieve:

1. Flux points tables
2. Grouped observational tables
3. Datasets ready for analysis

## 1. Retrieve Flux Points Table

Flux density measurements for a pulsar can be obtained as an Astropy table.
Each row corresponds to a measurement at a given frequency.

```python
from feupy.catalogs.pulsar_spectra import get_flux_points_table

# Retrieve the flux points table for a pulsar
table = get_flux_points_table("J1023+0038")

if table:
    print("Flux Points Table:")
    print(table)
```

## 2. Retrieve Grouped Tables

Flux measurements can also be retrieved as grouped tables.
These groups typically correspond to different observations, instruments,
or references.

Each table contains metadata describing the origin of the data.

```python
from feupy.catalogs.pulsar_spectra import get_tables

# Retrieve grouped tables for the pulsar
grouped_tables = get_tables("J1023+0038")

for group in grouped_tables:
    print("Grouped Table Metadata:")
    print(group.meta)
    print(group)
```

## 3. Retrieve Datasets

For analysis workflows, the pulsar measurements can be converted into
datasets compatible with analysis tools.

These datasets may contain spectral information ready for fitting or modeling.

```python
from feupy.catalogs.pulsar_spectra import get_datasets

# Retrieve datasets for the pulsar
datasets = get_datasets("J1023+0038")

print("Dataset Names:")
print(datasets.names)
```
