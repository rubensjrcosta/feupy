# FeuPy Source Catalog Examples

This notebook demonstrates how to work with source catalogs and source objects using FeuPy. The examples cover catalog access, source inspection, spectral information, flux points, and collections of sources.

````python
# Licensed under a 3-clause BSD style license - see LICENSE.rst

import gammapy

import feupy
from feupy.catalogs import (
    SourceCatalogExtraLHAASO,
    SourceCatalogLHAASO,
    SourceCatalogPSRCAT,
    SourceCatalogVERITASCygnus,
    SourceCatalogVTSCat,
)
from feupy.catalogs.utils import get_catalog_tag
from feupy.core.sources import Sources

print("FeuPy version:", feupy.__version__)
print("Gammapy version:", gammapy.__version__)
````

## LHAASO catalog

Load the catalog, inspect a source, and access its spectral model and flux-point information.

````python
lhaaso_catalog = SourceCatalogLHAASO()
lhaaso_source = lhaaso_catalog[0]

print(lhaaso_source)
print(lhaaso_source.info(info="all"))
````

````python
lhaaso_sky_model = lhaaso_source.sky_model()
lhaaso_spectral_model = lhaaso_source.spectral_model()

print("Sky model:")
print(lhaaso_sky_model)

print("\nSpectral model:")
print(lhaaso_spectral_model)
````

````python
lhaaso_source.flux_points
````

````python
lhaaso_source.flux_points_table
````

## Extra LHAASO catalog

The extra LHAASO catalog uses the same source interface, allowing the same workflow for model and flux-point access.

````python
extra_lhaaso_catalog = SourceCatalogExtraLHAASO()
extra_lhaaso_source = extra_lhaaso_catalog[0]

print(extra_lhaaso_source)
print(extra_lhaaso_source.sky_model())
print(extra_lhaaso_source.spectral_model())
````

````python
extra_lhaaso_source.flux_points
````

## VTSCat catalog

VTSCat sources provide additional observational and reference metadata.

````python
vts_catalog = SourceCatalogVTSCat()
vts_source = vts_catalog[0]

print(vts_source)
print(vts_source.info(info="all"))
print("Observation info:", vts_source.observation_info)
````

````python
vts_source.flux_points
````

````python
vts_source.flux_points_tables
````

````python
vts_source.reference_id_info
````

## PSRCAT catalog

````python
psrcat = SourceCatalogPSRCAT()
pulsar = psrcat[0]

print(pulsar)
print(pulsar.info(info="all"))
````

## VERITAS Cygnus catalog

````python
veritas_catalog = SourceCatalogVERITASCygnus()
veritas_source = veritas_catalog[5]

print(veritas_source)
print(veritas_source.info(info="all"))
````

````python
veritas_source.flux_points
````

## Working with `Sources`

Individual catalog sources can be grouped in a `Sources` collection for joint handling of labels, names, and positions.

````python
sources = Sources([lhaaso_source])

print("Labels:", sources.labels)
print("Names:", sources.names)
print("Positions:", sources.positions)
````

````python
source = sources[0]
name_label = (source.name, get_catalog_tag(source))
name_label
````

## Saving and loading source collections

`Sources` collections can be serialized to YAML. To avoid creating files when this example is executed, the write/read workflow is shown below without being run automatically:

```python
sources.write(filename="sources.yaml")

loaded_sources = Sources()
loaded_sources.read(filename="sources.yaml")
print(loaded_sources.names)
```
