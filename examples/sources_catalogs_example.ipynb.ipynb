# Feupy Sources Catalogs Examples

This notebook demonstrates how to work with source catalogs and source objects
using the `feupy` package.

We show how to:

- Load different source catalogs
- Access individual sources
- Inspect source information
- Retrieve flux points and spectral models
- Work with collections of sources

## Example 1: LHAASO Catalog

In this example we load the LHAASO catalog, access a source,
and inspect its spectral and flux point information.

```python
from feupy.catalog import SourceCatalogLHAASO

# Initialize the catalog
catalog = SourceCatalogLHAASO()

# Access the first source
first_source = catalog[0]

# Print basic information
print(first_source)

# Print detailed information
print(first_source.info(info="all"))

# Retrieve the sky model
sky_model = first_source.sky_model()
print(f"Sky model: {sky_model}")

# Access flux points
flux_points = first_source.flux_points
print(flux_points)

# Display the differential flux table
flux_table = first_source.flux_points_table
print(flux_table)

# Fit spectral model
spectral_model = first_source.spectral_model()

if spectral_model:
    print(f"Fitted Spectral Model: {spectral_model}")
else:
    print("Spectral model fitting was not successful or unsupported model type.")
```

## Example 2: 1LHAASO Catalog

```python
from feupy.catalog import SourceCatalogExtraLHAASO

catalog = SourceCatalogExtraLHAASO()

first_source = catalog[0]

print(first_source)

sky_model = first_source.sky_model()
print(f"Sky model: {sky_model}")

flux_points = first_source.flux_points
print(flux_points)


spectral_model = first_source.spectral_model()

if spectral_model:
    print(f"Fitted Spectral Model: {spectral_model}")
else:
    print("Spectral model fitting was not successful or unsupported model type.")
```

## Example 3: VTSCat Catalog

```python
from feupy.catalog import SourceCatalogVTSCat

catalog = SourceCatalogVTSCat()

source = catalog[0]

print(source)
print(source.info(info="all"))

obs_info = source.observation_info
print(f"Observation info: {obs_info}")

flux_points = source.flux_points
print(flux_points)

flux_tables = source.flux_points_tables
print(flux_tables)

source.reference_id_info
```

```python
catalog.table.colnames
```

## Example 4: Working with Multiple Sources

```python
from feupy.core.sources import Sources

sources = Sources([first_source])

sources.labels
sources.names
sources.positions
```

## Example 5: Save and Load Sources

```python
sources.write(filename="sources.yaml")

sources = Sources()
sources.read(filename="sources.yaml")

sources.names
```
