# Pulsar Spectrum Example

This notebook demonstrates how to access pulsar flux measurements and datasets using the FeuPy pulsar spectra interface.

The pulsar spectra data are based on the `pulsar_spectra` project:

https://all-pulsar-spectra.readthedocs.io/en/latest/

The examples below show how to retrieve flux-point tables, grouped observational tables, and datasets ready for analysis.

```python
# Licensed under a 3-clause BSD style license - see LICENSE.rst

import gammapy

import feupy
from feupy.catalogs.pulsar_spectra import (
    get_pulsar_flux_points_datasets,
    get_pulsar_flux_points_table,
    get_pulsar_flux_points_tables,
)

print("FeuPy version:", feupy.__version__)
print("Gammapy version:", gammapy.__version__)
```

## Retrieve a flux-points table

Flux-density measurements for a pulsar can be retrieved as an Astropy table. Each row corresponds to a measurement at a given frequency.

```python
pulsar_name = "J1023+0038"

flux_points_table = get_pulsar_flux_points_table(pulsar_name)
flux_points_table
```

## Retrieve grouped observational tables

Measurements can also be retrieved as grouped tables. The groups typically correspond to different observations, instruments, or references, and may contain metadata describing the data origin.

```python
grouped_tables = get_pulsar_flux_points_tables(pulsar_name)

for index, table in enumerate(grouped_tables, start=1):
    print(f"Group {index}")
    print("Metadata:", table.meta)
    print(table)
    print()
```

## Retrieve datasets

The pulsar measurements can be converted into datasets suitable for downstream spectral analysis and modeling.

```python
datasets = get_pulsar_flux_points_datasets(pulsar_name)

print("Dataset names:")
print(datasets.names)

datasets
```

## Summary

- convert pulsar data into datasets suitable for analysis.
