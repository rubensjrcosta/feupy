```python
import yaml
import numpy as np
from pulsar_spectra.catalogue import collect_catalogue_fluxes

# Collect the data
cat_dict = collect_catalogue_fluxes()

# Convert NumPy objects to plain Python types
def convert_numpy_objects(obj):
    """Recursively convert NumPy objects to plain Python types."""
    if isinstance(obj, np.ndarray):
        return obj.tolist()
    elif isinstance(obj, np.generic):
        return obj.item()  # Convert NumPy scalar to Python scalar
    elif isinstance(obj, dict):
        return {k: convert_numpy_objects(v) for k, v in obj.items()}
    elif isinstance(obj, list):
        return [convert_numpy_objects(x) for x in obj]
    elif isinstance(obj, tuple):
        return tuple(convert_numpy_objects(x) for x in obj)
    return obj

# Convert `cat_dict` to be YAML-safe
safe_cat_dict = convert_numpy_objects(cat_dict)

# Save the dictionary to YAML
with open("pulsar_spectra.yaml", "w") as yaml_file:
    yaml.dump(safe_cat_dict, yaml_file, default_flow_style=False)

```

```python
# Read the dictionary back from YAML
with open("pulsar_spectra.yaml", "r") as yaml_file:
    loaded_cat_dict = yaml.safe_load(yaml_file)
```

```python
# Licensed under a 3-clause BSD style license - see LICENSE.rst
"""Catalog utilities for pulsar spectra."""

import numpy as np
from astropy.units import Quantity
from astropy.table import Table, Column
from gammapy.datasets import Datasets, FluxPointsDataset
from gammapy.estimators import FluxPoints
from feupy.utils.units import Hz_to_eV, Jy_to_erg_by_cm2_s

# reference = 'https://doi.org/10.1038/s41586-021-03498-z'
# sed_type = "e2dnde"
# Bibcode = '2021Natur.594...33C'


catalog_table_meta = {
    "catalog": CATALOG_TAG,
    "SED_TYPE": sed_type,
    "comments": [f'reference: {reference}'],
    "Bibcode": Bibcode
}


def get_datasets_flux_points_pulsar_spectra(loaded_cat_dict, pulsar_Jname):
    """
    Generate flux points datasets for a pulsar from a catalog dictionary.

    Parameters
    ----------
    loaded_cat_dict : dict
        The catalog dictionary containing pulsar data.
    pulsar_Jname : str
        The pulsar J-name.

    Returns
    -------
    datasets : `gammapy.datasets.Datasets`
        Collection of flux points datasets.
    """
    try:
        # Extract pulsar data from catalog
        freqs, bands, fluxs, flux_errs, refs = loaded_cat_dict[pulsar_Jname]

        # Convert quantities to appropriate units
        freqs_MHz = Quantity(freqs, 'MHz')
        bands_MHz = Quantity(bands, 'MHz')
        fluxs_mJy = Quantity(fluxs, 'mJy')
        flux_errs_mJy = Quantity(flux_errs, 'mJy')

        # Initialize Astropy table with metadata
        table = Table()
        table.meta['Name'] = f'PSR {pulsar_Jname}'
#         table.meta['comments'] = ['Reference: https://pulsar-spectra.readthedocs.io']
        table['ref'] = np.array(refs, dtype='U20')
        table['ref'].description = "The reference label (in the format 'Author_year')"

        # Add computed columns for energy and flux
        table['e_ref'] = Column(
            data=Hz_to_eV(freqs_MHz),
            description='Energy of reference',
            unit='eV',
            format='.3e',
        )
        table['e2dnde'] = Column(
            data=Jy_to_erg_by_cm2_s(freqs_MHz, fluxs_mJy),
            description='Differential flux',
            unit='erg cm^-2 s^-1',
            format='.3e',
        )
        table['e2dnde_err'] = Column(
            data=Jy_to_erg_by_cm2_s(freqs_MHz, flux_errs_mJy),
            description='Differential flux uncertainties',
            unit='erg cm^-2 s^-1',
            format='.3e',
        )

        # Group by reference and create datasets
        datasets = Datasets()
        grouped_data = table.group_by('ref')

        for group in grouped_data.groups:
            label = group['ref'][0].replace('_', '-')  # Generate label from reference
            flux_points = FluxPoints.from_table(
                table=group[['e_ref', 'e2dnde', 'e2dnde_err']],
                sed_type='e2dnde',
            )
            dataset = FluxPointsDataset(data=flux_points, name=label)
            datasets.append(dataset)

    except KeyError as error:
        print(f"KeyError: Pulsar J-name '{pulsar_Jname}' not found in catalog.")
    except Exception as error:
        print(f"Error processing pulsar '{pulsar_Jname}': {error}")

    return datasets
```
